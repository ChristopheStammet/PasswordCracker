```python
# <------- WICHTIG!!! Drücke hier auf Play
#
# Verändere an diesem Code bitte überhaupt nichts!

import hashlib
import itertools
from tqdm.notebook import tqdm

def hash(input_string):
    return hashlib.md5(input_string.encode()).hexdigest()

def get_name_by_hash(d, hash_value):
    for name, h in d.items():
        if h == hash_value:
            return name
    return None  # If not found

def generate_strings(chars, length):
    """Generates all possible strings of a given length using a character list."""
    for combination in itertools.product(chars, repeat=length):
        yield "".join(combination)

dataleak_dict = {
    "Alex": "a08372b70196c21a9229cf04db6b7ceb",
    "Béatrice": "8fe4c11451281c094a6578e6ddbf5eed",
    "Christophe": "2b9ed9314c9b75f085fcf7356337d30b",
    "Daniela": "6c96314949a6b7cd3635c4ea904bb7d0",
    "Elisabeth": "a384b6463fc216a5f8ecb6670f86456a",
    "Fernanda": "25d55ad283aa400af464c76d713c07ad",
    "Gilles": "5d793fc5b00a2348c3fb9ab59e5ca98a",
    "Hannah": "846519378c9dedfaadf6c21bbebf1f4c",
    "Ilsa": "735b90b4568125ed6c3f678819b6e058",
    "Jhempi": "5993d7a73d9f9a694e411ba0788cfe2d",
    "Klara": "4fc963e213bba362778f5c175eb4d5ff",
    "Louis": "7f320b406a0956586fcc21c9f18e9180",
    "Maren": "c2f3f489a00553e7a01d369c103c7251",
    "Nico": "5f4dcc3b5aa765d61d8327deb882cf99",
    "Olivia": "bfa08887053e473ace6f22633348634e",
    "Pedro": "8acef4050a09ce337a04186afd44ed33",
    "Quentin": "1a36591bceec49c832079e270d7e8b73",
    "Robert": "48e6c9d963b4ec1c7507c505d577a6ad",
    "Svenja": "0aed5d740d7fab4201e885019a36eace",
    "Tom": "f1174e62d60a92010c4a72fe87805ae1",
    "Ulrike": "244e457150727f77a5f07ca17969382f",
    "Victor": "4de8e33e649c6ee317c7937804b63fc1",
    "Wilhelmina": "9dd4e461268c8034f5c8564e155c67a6",
    "Xenia": "38b77d171bec2ddcadcbf434355b4184",
    "Yves": "06a82af8f6f22be62676d5ff0a4de161",
    "Zelda": "c4657b50cfed11e0005ec752fa01a651"
}

dataleak = list(dataleak_dict.values())
print("Alles bereit! Du kannst loslegen!")
```

## Bruteforce - alles ausprobieren


> Der nächste Schritt ist es, alle möglichen Kombinationen von kleinen Buchstaben einer bestimmten Länge "testlength" generieren zu lassen und zu schauen, ob sie sich in den Daten befindet.

ACHTUNG: Eine zu hohe Länge wird ewig dauern!



```python
charlist = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']
# ===== Verändere hier die Passwortlänge ==============
testlength = 3
# ============================

total_permutations = len(charlist) ** testlength

for test_password in tqdm(generate_strings(charlist, testlength), total=total_permutations):
    password_hash = hash(test_password)
    if password_hash in dataleak_dict.values():
        name = get_name_by_hash(dataleak_dict, password_hash)
        print(f"PassworT gefunden: Dies ist {name}'s Passwort: {test_password}")
```

# Angriff Nummer 4
## Gezielte Veränderungen



> Viele Passwörter, die Sonderzeichen enthalten müssen, sind immer auf die gleiche Weise verändert, indem z.B. ein 'e' zu einem '€' oder einer '3', ein 's' zu einem '$' oder ein 'a' zu einem '@' werden.

Versuche hier, einige solche Veränderungen (substitutions) zu definieren und lasse den Computer dann sämtliche Veränderungen austesten.



```python
# ==== Füge mögliche Veränderungen in diese Liste hinzu ====
# ==== Links vom : ist das zu ersetzende Symbol
# ==== Rechts vom : ist die Liste der möglichen Veränderungen
# ==== Setze das Symbol der Veränderung zwischen die '' oder füge weitere Zeilen hinzu

substitutions = {
    'a': ['a', '@'],
    'e': ['e', '€', '3'],
}

# ==== Hier kannst du natürlich auch die Liste der Charaktere sowie die Länge des Passwortes anpassen
charlist = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']
testlength = 4
# ==========================

def apply_substitutions(word, subs):
    """Generate all variants of word by substituting each character with all allowed substitutions."""
    # For each character, get the list of substitutions or the char itself
    lists_of_chars = [subs.get(c, [c]) for c in word]
    # Cartesian product of these lists gives all variants
    for variant_tuple in itertools.product(*lists_of_chars):
        yield "".join(variant_tuple)

def generate_strings(chars, length):
    for combo in itertools.product(chars, repeat=length):
        yield "".join(combo)

total_base = len(charlist) ** testlength

# Since substitution expansions multiply variants, total variants is much larger — warn pupils

for base_word in tqdm(generate_strings(charlist, testlength), total=total_base):
    for variant in apply_substitutions(base_word, substitutions):
      password_hash = hash(variant)
      if password_hash in dataleak_dict.values():
          name = get_name_by_hash(dataleak_dict, password_hash)
          print(f"Password gefunden: {variant} ist das Passwort von {name}")
```
